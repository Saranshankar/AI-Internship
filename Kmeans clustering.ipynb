```python
q=df.iloc[:,3:].values
q
```

```python
from sklearn.cluster import KMeans
iner=[]
for i in range(1,11):
  kmeans=KMeans(n_clusters=i,init='k-means++',max_iter=300,n_init=10)
  kmeans.fit(q)
  iner.append(kmeans.inertia_)
plt.plot(iner)
```

```python
kmeans=KMeans(n_clusters=4,init='k-means++')
y_pred=kmeans.fit_predict(q)
```

```python
y_pred
```

```python
kmeans.inertia_
```

```python
kmeans.labels_
```

```python
plt.scatter(q[y_pred==0,0],q[y_pred==0,1],color='green',label='cluster1')
plt.scatter(q[y_pred==1,0],q[y_pred==1,1],color='blue',label='cluster1')
plt.scatter(q[y_pred==2,0],q[y_pred==2,1],color='orange',label='cluster1')
plt.scatter(q[y_pred==3,0],q[y_pred==3,1],color='red',label='cluster1')
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],color="pink",marker='>')
```

```python
from sklearn.cluster import DBSCAN
db=DBSCAN(eps=10,min_samples=3,metric='euclidean')
model=db.fit(q)
```

```python
import matplotlib.pyplot as plt

# Get cluster labels
labels = model.labels_

# Plot
plt.figure(figsize=(8,6))
plt.scatter(q[:,0], q[:,1], c=labels, cmap='rainbow', s=100)

plt.title("DBSCAN Clustering")
plt.xlabel("X")
plt.ylabel("Y")
plt.colorbar(label="Cluster")
plt.show()
```
