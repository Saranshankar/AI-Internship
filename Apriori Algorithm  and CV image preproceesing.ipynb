```python
import numpy as np
transaction = []
for i in range(len(df)):
    for j in range(20):
        value = df.values[i, j]
        if value is not None and not (isinstance(value, float) and np.isnan(value)):
            transaction.append([str(value)])
transaction
```

```python
!pip install apyori
```

```python
from apyori import apriori
rules=apriori(transactions=transaction,min_support=0.003,min_confidence=0.2,min_lift=1.0,min_length=2)
results=list(rules)
results
```

```python
import pandas as pd

results = list(results)

rules_df = pd.DataFrame([record._asdict() for record in results])

rules_df
```
