```python
"""
🔹 What is Power Transformer?

It is used to:

Make data more normal (Gaussian-like)
Reduce skewness
Stabilize variance

👉 More powerful than log transform because it automatically finds the best transformation

🔹 Types
Yeo-Johnson → works with zero & negative values ✅
Box-Cox → works only with positive values ❌ (no zero/negative)

"""
# Step 1: Detect outliers FIRST
from scipy.stats import zscore
import numpy as np

d = df1.select_dtypes(include='number')
d = d.fillna(d.mean())

z = np.abs(zscore(d['Age']))
outliers = z > 2

print(outliers)
from sklearn.preprocessing import PowerTransformer

pt = PowerTransformer(method='yeo-johnson')
power_t = pt.fit_transform(d[['Age']])
```

```python
from scipy.stats import zscore
import numpy as np

d = df1.select_dtypes(include='number')
d = d.fillna(d.mean())

z = np.abs(zscore(d['Age']))
outliers_before = z > 2

print(outliers_before)
from sklearn.preprocessing import PowerTransformer

pt = PowerTransformer(method='box-cox')

boxcox_data = pt.fit_transform((d[['Age']] + 1))  # +1 to avoid zero
```

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

d = df1.select_dtypes(include='number')

# Handle missing values
d = d.fillna(d.mean())

scaler = StandardScaler()
scaled_data = scaler.fit_transform(d)

scaled_df = d.copy()
scaled_df[:] = scaled_data
outliers = np.abs(scaled_df) > 2.5
print(outliers)
```
