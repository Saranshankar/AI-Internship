```python
model=SentenceTransformer("paraphrase-albert-small-v2")
doc_embedding=model.encode(docs).astype("float32")
```

```python
doc_embedding.shape
```

```python
dimension=doc_embedding.shape[1]
index=faiss.IndexFlatL2(dimension)
index.add(doc_embedding)
```

```python
ques="what is LlamaIndex  ?"
ques_embedding=model.encode([ques]).astype("float32")
distance,indices=index.search(ques_embedding,2)
retrieve=[docs[i] for i in indices[0]]
for ans in retrieve:
  print("*",ans)
```

```python
from openai import OpenAI
```

```python
from groq import Groq
client = Groq(api_key="your key")
context = "\n".join(retrieve)
prompt = f"""
Answer the question based on the context below.
Context:
{context}
Question:
{ques}
Answer:
"""
response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "user", "content": prompt}
    ],
    temperature=0.2,
)

print("Final Answer:\n")
print(response.choices[0].message.content)
```
