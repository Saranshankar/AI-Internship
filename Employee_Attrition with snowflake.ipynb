```python
import matplotlib.pyplot as plt
import math
col = df.select_dtypes(include=['int64', 'float64']).columns

n = len(col)
rows = math.ceil(n / 4)
plt.figure(figsize=(15, rows * 4))

for i in range(n):
    plt.subplot(rows, 4, i + 1)
    plt.hist(df[col[i]])
    plt.title(col[i])

plt.tight_layout()
plt.show()
```
