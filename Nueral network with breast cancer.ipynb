```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
```

```python
breast_cancer_dataset=sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer_dataset)
```

```python
df = pd.DataFrame(
    data=breast_cancer_dataset.data,
    columns=breast_cancer_dataset.feature_names
)
df['target']=breast_cancer_dataset.target
df
```

```python
df.info()
```

```python
c = (df['target'] == 0).sum()
p = (df['target'] == 1).sum()

print(c)
print(p)
```

```python
X=df.drop('target',axis=1)
y=df['target']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3)
```

```python
from sklearn.preprocessing import StandardScaler
x_scaled=StandardScaler()
train_data=x_scaled.fit_transform(X_train,y_train)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Standardize training data
X_train_scaled = scaler.fit_transform(X_train)

# Standardize test data
X_test_scaled = scaler.transform(X_test)
```

```python
import tensorflow as tf
from tensorflow import keras
```

```python
model= keras.Sequential([
    keras.layers.Flatten(input_shape=(30,)),
    keras.layers.Dense(20,activation='relu'),
    keras.layers.Dense(2,activation='sigmoid')
])
```

```python
model
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
history=model.fit(X_train_scaled,y_train,validation_split=0.1,epochs=40)
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title("accuracy comparision")
plt.ylabel('accuracy')
plt.xlabel('epoch')
plt.legend(['training data','validation data'],loc='lower right')
```
