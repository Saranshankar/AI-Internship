```python
from transformers import pipeline
classifier=pipeline("sentiment-analysis")
classifier("today i fallen in valley")
```

```python
classifier=pipeline("zero-shot-classification")
classifier("toady i learning in new ",
      candidate_labels=["chatgpt","claude","geekforgeeks"])
```

```python
classi=pipeline("text-generation")
classi("today i playing game was")
```

```python
pipe = pipeline("text-generation", model="Qwen/Qwen3-0.6B")
pipe("today i playing game was",max_length=20,num_return_sequences=2)
```

```python
pipe=pipeline("fill-mask",top_k=2)
pipe("today i working at <mask> company")
```

```python
pipee = pipeline("ner", aggregation_strategy="simple")
pipee("vaibav suriyavanshi is a young cricket player")
```

```python
from transformers import pipeline

pipep = pipeline(
    task="question-answering",
    model="distilbert-base-uncased-distilled-squad"
)

result = pipep(
    question="What is the capital of India?",
    context="The capital of India is Delhi."
)

print(result)
```

```python
from transformers import pipeline

pipe = pipeline(
    "translation",
    model="facebook/nllb-200-distilled-600M"
)

pipe(
    "வணக்கம், நீங்கள் எப்படி இருக்கிறீர்கள்?",
    src_lang="tam_Taml",
    tgt_lang="eng_Latn"
)
```

```python
from transformers import pipeline

pipp = pipeline("summarization")

text = """
Grab a cup of coffee and settle in for the incredible true story of the most successful art heist in history, centered on a masterpiece that vanished into the Sicilian underworld. In 1610, the legendary Italian master Caravaggio completed a breathtaking, towering altarpiece titled Nativity with St. Francis and St. Lawrence, which hung peacefully for nearly 350 years inside a historic chapel in Palermo, Sicily. That peace shattered on the dark, rainy night of October 17, 1969, when amateur thieves slipped into the church, crudely sliced the massive canvas right out of its frame, and rolled it up to steal away into the night. The Italian authorities immediately launched a massive, decades-long investigation, but the trail quickly led straight into the secretive and brutal world of the Sicilian Mafia. Over the years, terrifying and tragic rumors emerged from mob informants: some claimed the painting was used as a literal floor rug by a ruthless crime boss, others swore it was stashed in a damp barn where it was destroyed by rats and eventually burned, and some insisted it survived intact as a multi-million-dollar collateral piece for international drug and weapon deals. Today, a high-tech replica hangs in the empty frame in Palermo, while Caravaggio's original masterwork remains on the FBI's Top Ten Art Crimes list, valued at over $20 million—a haunting, unsolved ghost story of a priceless cultural treasure lost to the shadows.
"""

result = pipp(text)
print(result)
```
